#### mengapa data mentah dan data bersih dipisah direktori, dan apa arti replikasi = 1 pada VM pseudo-distributed dibanding klaster nyata.
Data mentah dan data bersih kita pisah direktori untuk menjaga keaslian data, ini sangat penting karena untuk membuktikan bahwa yang dipakai itu benar benar data dari tidarmart yang belum diolah, kita bisa membuktikannya melalui data mentah.
Pemisahan ini juga mempermudah kita jika setelah dilakukannya pembersihan data terdapat kesalahan dalam pembersihannya, jika kita menggabungkannya maka kita telah kehilangan data mentah tadi (karena tertimpa) dan pembersihan tidak berjalan dengan baik(karena tidak bisa diulang). Jika kita menggunakan data yang terpisah dan terjadi kesalahan kita dapat membersihkannya ulang karena data mentahnya masih utuh pada direktori yang berbeda.

Replikasi = 1 berarti ia menyimpan data atau mereplikasinya hanya sebanyak 1 kali. Pada klaster nyata, replikasi berguna untuk menjaga data, jika satu mesin mati mendadak, kita masih punya salinannya di mesin lain.
Kemudian VM pseudo-distributed adalah cara kita menjalankan sistem terdistribusi diatas satu mesin tunggal tetapi layanannya berjalan sebagai proses terpisah seperti cara kerja multi klaster. DIbanding klaster nyata pseudo-distributed tidak perlu menyiapkan banyak mesin fisik untuk melakukan replikasi data, cukup dengan 1 VM.
Di VM pseudo-distributed, karena tiap blok hanya punya 1 salinan di 1 mesin, jika VM atau DataNode-nya bermasalah maka datanya bisa hilang atau tidak bisa diakses, karena tidak ada salinan di mesin lain.


## Tahap 2
- Hitung jumlah baris, jumlah order_id unik, persentase nilai kosong per kolom, jumlah harga   negatif, distribusi qty (temukan outlier dengan kuantil, jenis nya bebas), dan variasi penulisan kota.
- Sajikan hasil profiling dalam satu tabel ringkasan “sebelum pembersihan”.
- Untuk profil pelanggan dan produk: tunjukkan bahwa atribut produk bervariasi me

```python
from pyspark.sql import SparkSession

spark = (SparkSession.builder
    .appName("TidarMart")
    .master("local[2]")
    .config("spark.driver.memory", "1536m")
    .config("spark.sql.shuffle.partitions", "4")
    .config("spark.sql.adaptive.enabled", "true")
    .getOrCreate())

spark.sparkContext.setLogLevel("ERROR")
print("SparkSession siap. Versi Spark:", spark.version)
```

```python
from pyspark.sql.functions import col, count, when, round as _round

# membaca file
df_transaksi = spark.read.csv(
    "hdfs://localhost:9000/tidarmart/raw/transaksi",
    header=True, inferSchema=True)
df_produk = spark.read.json(
    "hdfs://localhost:9000/tidarmart/raw/produk")
df_pelanggan = spark.read.json(
    "hdfs://localhost:9000/tidarmart/raw/pelanggan")

# skema dataframe
print(f"Jumlah baris df_transaksi: {df_transaksi.count()}")
df_transaksi.printSchema()
print(f"Jumlah baris df_produk: {df_produk.count()}")
df_produk.printSchema()
print(f"Jumlah baris df_pelanggan: {df_pelanggan.count()}")
df_pelanggan.printSchema()
print()

# menghitung baris,jumlah order_id unik, jumlah harga negatif
baris_transaksi = df_transaksi.count()
baris_produk = df_produk.count()
baris_pelanggan = df_pelanggan.count()
order_id_unik = df_transaksi.select("order_id").distinct().count()
harga_negatif = df_transaksi.filter(col("harga_satuan") < 0).count()

print("jumlah baris transaksi:", baris_transaksi)
print("jumlah baris produk:", baris_produk)
print("jumlah baris pelanggan:", baris_pelanggan)
print("jumlah order id unik:", order_id_unik)
print("jumlah harga negatif:", harga_negatif )
```

```python
# cek_kosong = df_transaksi.select([
#     _round(count(when(col(c).isNull(), c)))
#     for c in df_transaksi.columns
# ])
# cek_kosong.show(vertical=True)

# menghitung nilai persentase perkolom
persen_kosong = df_transaksi.select([
    _round(count(when(col(c).isNull(), c)) / baris_transaksi * 100, 2).alias(c)
    for c in df_transaksi.columns
])
persen_kosong.show(vertical=True)
```
